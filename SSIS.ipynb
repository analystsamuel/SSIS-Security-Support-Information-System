```python
import pandas as pd

df_incidents = pd.read_csv(r"C:\Users\Samuel\Documents\incident_df.csv")
df_sites = pd.read_csv(r"C:\Users\Samuel\Documents\ssis_sites.csv")
df_employees = pd.read_csv(r"C:\Users\Samuel\Documents\employee_df.csv")

print("Incidents columns:", df_incidents.columns.tolist())
print("Sites columns:", df_sites.columns.tolist())
print("Employees columns:", df_employees.columns.tolist())
```

```python
df_incidents['location'] = df_incidents['location'].astype(str).str.strip().str.lower()
df_incidents['incident_type'] = df_incidents['incident_type'].astype(str).str.strip().str.lower()
df_incidents['incident_datetime'] = pd.to_datetime(df_incidents['incident_date_time'], format='mixed', errors='coerce')
print(df_incidents[['location', 'incident_type', 'incident_datetime']].head())
```

```python
print(df_incidents['incident_date_time'].head(10))
```

```python
df_incidents['start_time_str'] = df_incidents['incident_date_time'].astype(str).str.split('-').str[0].str.strip()
df_incidents['incident_datetime'] = pd.to_datetime(df_incidents['start_time_str'], format='mixed', errors='coerce')
if 'date of report' in df_incidents.columns:
    df_incidents['report_date'] = pd.to_datetime(df_incidents['date of report'], format='mixed', errors='coerce')
    df_incidents['incident_datetime'] = df_incidents['incident_datetime'].fillna(df_incidents['report_date'])
print(df_incidents[['incident_date_time', 'incident_datetime']].head(10))
```

```python
import numpy as np

df_sites = pd.read_csv(r"C:\Users\Samuel\Documents\ssis_sites.csv")
df_employees = pd.read_csv(r"C:\Users\Samuel\Documents\employee_df.csv")

df_sites['location'] = df_sites['location'].astype(str).str.strip().str.lower()

df_incidents = pd.merge(df_incidents, df_sites[['site_id', 'location']], on='location', how='left')

df_incidents['site_id'] = df_incidents['site_id'].fillna(1).astype(int)

emp_ids = df_employees['EmpID'].dropna().unique()
df_incidents['employee_id'] = np.random.choice(emp_ids, size=len(df_incidents))

print(df_incidents[['site_id', 'employee_id', 'location']].head(10))
```

```python
import numpy as np

severities = ['low', 'medium', 'high']
df_incidents['severity'] = np.random.choice(severities, size=len(df_incidents), p=[0.5, 0.35, 0.15])
df_incidents['response_time_minutes'] = np.random.randint(5, 46, size=len(df_incidents))
statuses = ['closed', 'under investigation', 'open']
df_incidents['status'] = np.random.choice(statuses, size=len(df_incidents), p=[0.8, 0.15, 0.05])
final_cols = ['site_id', 'employee_id', 'incident_datetime', 'incident_type', 'severity', 'response_time_minutes', 'status']
df_final = df_incidents[final_cols]
df_final.to_csv('cleaned_incidents.csv', index=False)
print(df_final.head())
```
